# Level 1: Python Variables and Data Types

This notebook introduces the most basic Python ideas: variables, numbers, text, and simple calculations.

A variable is a name that stores a value.

## 1. Create variables

```python
gene_name = "TP53"
chromosome = 17
expression_value = 120.5

gene_name, chromosome, expression_value
```

## 2. Check data types

`type()` tells you what kind of value a variable contains.

```python
type(gene_name)
```

```python
type(chromosome)
```

```python
type(expression_value)
```

## 3. Do calculations

```python
sample_1 = 100
sample_2 = 140
sample_3 = 120

mean_expression = (sample_1 + sample_2 + sample_3) / 3
mean_expression
```

## 4. Combine text and variables

An f-string lets you place variables inside text.

```python
message = f"The gene {gene_name} is on chromosome {chromosome}."
message
```
