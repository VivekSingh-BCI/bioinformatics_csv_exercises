# Level 8: SciPy Statistics and t-tests

`SciPy` contains many scientific computing tools.

In this notebook you will use `scipy.stats` to compare two small groups of expression values.

```python
import numpy as np
from scipy import stats
```

## 1. Create two groups of values

These example values could represent gene expression in control and disease samples.

```python
control = np.array([10.2, 11.1, 9.8, 10.5, 10.9])
disease = np.array([14.8, 15.2, 13.9, 16.1, 15.5])

control, disease
```

## 2. Calculate means

```python
control.mean(), disease.mean()
```

## 3. Run an independent t-test

A t-test can be used to compare the means of two groups.

For this practice exercise, focus on running the code and reading the output.

```python
test_result = stats.ttest_ind(control, disease)
test_result
```

## 4. Access the statistic and p-value

```python
test_result.statistic
```

```python
test_result.pvalue
```

## 5. Make a simple decision

A common threshold is 0.05. This does not prove biology, but it is useful coding practice.

```python
if test_result.pvalue < 0.05:
    print("The groups are different using p < 0.05.")
else:
    print("The groups are not different using p < 0.05.")
```
