# Level 13: Filtering Genomic Variants

In this notebook you will filter a variant table.

Variant tables often contain many rows. Bioinformaticians usually apply filters to focus on variants with enough evidence or higher predicted impact.

```python
import pandas as pd
```

## 1. Load the variants CSV

```python
variants = pd.read_csv("data/variants.csv")
variants
```

## 2. Filter by read depth

Read depth is the number of sequencing reads supporting a genomic position. Higher depth usually gives more confidence.

```python
high_depth = variants[variants["read_depth"] >= 50]
high_depth
```

## 3. Filter by predicted impact

```python
high_impact = variants[variants["impact"] == "HIGH"]
high_impact
```

## 4. Combine filters

Use `&` for AND. Each condition must be inside parentheses.

```python
important_variants = variants[(variants["read_depth"] >= 50) & (variants["impact"] == "HIGH")]
important_variants
```

## 5. Sort variants

Sorting helps you see the largest or smallest values first.

```python
variants.sort_values("allele_frequency", ascending=False)
```
