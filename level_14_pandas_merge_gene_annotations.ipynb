# Level 14: Merging Gene Tables

In this notebook you will combine two CSV files:

- gene expression values
- gene annotations

This is common in bioinformatics because useful information is often stored in separate files.

```python
import pandas as pd
```

## 1. Load both CSV files

```python
expression = pd.read_csv("data/gene_expression.csv")
annotations = pd.read_csv("data/gene_annotations.csv")
```

```python
expression.head()
```

```python
annotations.head()
```

## 2. Merge the tables

Both tables have a column called `gene_name`. We can use this column to join them.

```python
merged = pd.merge(expression, annotations, on="gene_name")
merged
```

## 3. Calculate mean expression after merging

```python
merged["mean_expression"] = merged[["sample_A", "sample_B", "sample_C"]].mean(axis=1)
merged[["gene_name", "pathway", "mean_expression"]]
```

## 4. Group by pathway

Grouping lets us summarise rows that belong to the same category.

```python
pathway_summary = merged.groupby("pathway")["mean_expression"].mean().reset_index()
pathway_summary
```
