# Level 10: Simple Clustering with SciPy

Clustering groups similar samples or genes together.

This notebook introduces hierarchical clustering using a small expression matrix.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
```

## 1. Create a small expression matrix

Rows are samples. Columns are genes.

```python
sample_names = ["Control_1", "Control_2", "Disease_1", "Disease_2"]

expression = np.array([
    [10, 12, 9, 11],
    [11, 13, 10, 12],
    [30, 35, 28, 32],
    [29, 34, 27, 33]
])

expression
```

## 2. Run hierarchical clustering

`linkage()` calculates how samples should be grouped based on their values.

```python
cluster_result = linkage(expression, method="average")
cluster_result
```

## 3. Draw a dendrogram

A dendrogram is a tree-like plot showing which samples are most similar.

```python
plt.figure(figsize=(7, 4))
dendrogram(cluster_result, labels=sample_names)
plt.ylabel("Distance")
plt.title("Sample clustering")
plt.tight_layout()
plt.show()
```

## 4. Try a different method

Different methods can give slightly different clustering results.

```python
complete_result = linkage(expression, method="complete")

plt.figure(figsize=(7, 4))
dendrogram(complete_result, labels=sample_names)
plt.ylabel("Distance")
plt.title("Sample clustering using complete linkage")
plt.tight_layout()
plt.show()
```
