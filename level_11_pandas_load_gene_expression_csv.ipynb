# Level 11: Loading Gene Expression CSV Data

In this notebook you will learn how to load a CSV file using `pandas`.

The data contains expression values for several genes across three samples.

## 1. Import pandas

`pandas` is a Python library for working with tables.

```python
import pandas as pd
```

## 2. Load the CSV file

`read_csv()` reads a CSV file and stores it as a table called a DataFrame.

```python
expression = pd.read_csv("data/gene_expression.csv")
expression
```

## 3. Inspect the data

These commands help you understand the table before doing any analysis.

```python
expression.head()
```

```python
expression.shape
```

```python
expression.columns
```

## 4. Select columns

Use square brackets to select one column from the table.

```python
expression["gene_name"]
```

## 5. Calculate the average expression

Here we calculate the mean expression for each gene across the three samples.

```python
expression["mean_expression"] = expression[["sample_A", "sample_B", "sample_C"]].mean(axis=1)
expression
```
