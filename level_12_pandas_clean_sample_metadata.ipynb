# Level 12: Cleaning Sample Metadata

In this notebook you will load sample metadata and deal with missing values.

Metadata describes samples. In bioinformatics, metadata can include tissue type, disease status, age, sequencing depth, and other information.

```python
import pandas as pd
```

## 1. Load the metadata CSV

```python
metadata = pd.read_csv("data/sample_metadata.csv")
metadata
```

## 2. Check for missing values

`isna()` checks whether a value is missing. `sum()` counts how many missing values are in each column.

```python
metadata.isna().sum()
```

## 3. Fill missing numeric values

For this exercise, we will fill missing `age` and `sequencing_depth` values using the column median.

```python
metadata["age"] = metadata["age"].fillna(metadata["age"].median())
metadata["sequencing_depth"] = metadata["sequencing_depth"].fillna(metadata["sequencing_depth"].median())
metadata
```

## 4. Count samples by condition

`value_counts()` counts how many times each value appears.

```python
metadata["condition"].value_counts()
```

## 5. Save the cleaned table

`to_csv()` writes a DataFrame to a new CSV file. `index=False` prevents pandas from adding an extra row number column.

```python
metadata.to_csv("data/sample_metadata_cleaned.csv", index=False)
```
