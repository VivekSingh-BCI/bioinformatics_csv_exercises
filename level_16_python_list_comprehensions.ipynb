# Level 16: Python List Comprehensions

A list comprehension is a short way to create a new list from an existing list.

This is useful in bioinformatics when processing many genes, samples, or sequence lengths.

## 1. Start with a normal loop

```python
expression_values = [10, 25, 80, 45, 100]

doubled = []
for value in expression_values:
    doubled.append(value * 2)

doubled
```

## 2. Write the same idea as a list comprehension

```python
doubled = [value * 2 for value in expression_values]
doubled
```

## 3. Filter values using a list comprehension

```python
high_values = [value for value in expression_values if value > 50]
high_values
```

## 4. Work with gene names

```python
genes = ["tp53", "brca1", "egfr", "braf"]
upper_genes = [gene.upper() for gene in genes]
upper_genes
```

## 5. Calculate sequence lengths

```python
sequences = ["ATGC", "ATGCGG", "AT", "ATGCCCTA"]
sequence_lengths = [len(sequence) for sequence in sequences]
sequence_lengths
```
