# Level 2: Python Lists and Indexing

A list stores multiple values in one variable.

Bioinformatics data often contains lists of genes, samples, sequences, or measurements.

## 1. Create a list

```python
genes = ["TP53", "BRCA1", "BRCA2", "EGFR", "BRAF"]
genes
```

## 2. Select items from a list

Python starts counting at 0. The first item is at index 0.

```python
genes[0]
```

```python
genes[2]
```

## 3. Count items

```python
len(genes)
```

## 4. Add and remove items

```python
genes.append("MYC")
genes
```

```python
genes.remove("EGFR")
genes
```

## 5. Numeric lists

```python
read_counts = [12000, 15000, 18000, 11000, 22000]

minimum_reads = min(read_counts)
maximum_reads = max(read_counts)
total_reads = sum(read_counts)

minimum_reads, maximum_reads, total_reads
```
