# Level 6: Python Dictionaries and Records

A dictionary stores data as key-value pairs.

Dictionaries are useful for storing information about genes, samples, variants, or patients.

## 1. Create a dictionary

```python
gene = {
    "name": "TP53",
    "chromosome": 17,
    "start": 7661779,
    "end": 7687550
}

gene
```

## 2. Access values by key

```python
gene["name"]
```

```python
gene["chromosome"]
```

## 3. Add a new key

```python
gene["pathway"] = "DNA repair"
gene
```

## 4. A list of dictionaries

This structure is common when each item has several pieces of information.

```python
samples = [
    {"sample_id": "S01", "condition": "control", "reads": 12000},
    {"sample_id": "S02", "condition": "disease", "reads": 18000},
    {"sample_id": "S03", "condition": "control", "reads": 15000}
]

samples
```

## 5. Loop through records

```python
for sample in samples:
    print(sample["sample_id"], sample["condition"], sample["reads"])
```
