# Level 19: Simple Codon Translation

A codon is a group of three DNA bases.

In this notebook you will translate a short DNA sequence into amino acid letters using a small codon table.

## 1. Create a small codon table

```python
codon_table = {
    "ATG": "M",
    "TTT": "F",
    "TTC": "F",
    "GAA": "E",
    "GAG": "E",
    "TAA": "*",
    "TAG": "*",
    "TGA": "*"
}

codon_table
```

## 2. Select codons from a DNA sequence

```python
dna = "ATGGAGTTTTAA"

first_codon = dna[0:3]
second_codon = dna[3:6]

first_codon, second_codon
```

## 3. Translate one codon

```python
codon_table[first_codon]
```

## 4. Loop through codons

```python
for position in range(0, len(dna), 3):
    codon = dna[position:position + 3]
    print(codon)
```

## 5. Translate a short sequence

```python
protein = ""

for position in range(0, len(dna), 3):
    codon = dna[position:position + 3]
    amino_acid = codon_table[codon]
    protein = protein + amino_acid

protein
```

## 6. Put the code into a function

```python
def translate_dna(dna):
    protein = ""
    for position in range(0, len(dna), 3):
        codon = dna[position:position + 3]
        protein = protein + codon_table[codon]
    return protein

translate_dna("ATGGAGTTTTAA")
```
