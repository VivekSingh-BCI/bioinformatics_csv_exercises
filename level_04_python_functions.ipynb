# Level 4: Python Functions

A function is a reusable block of code.

Functions help avoid repeating the same code many times.

## 1. Define a simple function

```python
def describe_gene(gene_name):
    message = f"The gene name is {gene_name}."
    return message
```

```python
describe_gene("TP53")
```

## 2. A function with a calculation

```python
def calculate_mean(value_1, value_2, value_3):
    mean_value = (value_1 + value_2 + value_3) / 3
    return mean_value
```

```python
calculate_mean(100, 120, 140)
```

## 3. A function with a condition

```python
def pass_depth_filter(read_depth):
    if read_depth >= 50:
        return "Pass"
    else:
        return "Fail"
```

```python
pass_depth_filter(80)
```

```python
pass_depth_filter(25)
```

## 4. Use a function inside a loop

```python
depths = [20, 50, 65, 10, 90]

for depth in depths:
    result = pass_depth_filter(depth)
    print(depth, result)
```
