# Level 20: Reading Bioinformatics Sequence Files

Bioinformatics data is often stored in text files.

This notebook introduces simple file reading using a small FASTA file.

## 1. Read all lines from a file

```python
with open("data/example_sequences.fasta") as file:
    lines = file.readlines()

lines
```

## 2. Remove newline characters

```python
clean_lines = []

for line in lines:
    clean_lines.append(line.strip())

clean_lines
```

## 3. Separate headers and sequences

```python
headers = []
sequences = []

for line in clean_lines:
    if line.startswith(">"):
        headers.append(line)
    else:
        sequences.append(line)

headers, sequences
```

## 4. Calculate sequence lengths

```python
for sequence in sequences:
    print(sequence, len(sequence))
```

## 5. Calculate GC percentage for each sequence

```python
def gc_percent(sequence):
    gc_count = sequence.count("G") + sequence.count("C")
    return gc_count / len(sequence) * 100

for header, sequence in zip(headers, sequences):
    print(header, round(gc_percent(sequence), 2))
```
