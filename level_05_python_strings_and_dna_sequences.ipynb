# Level 5: Python Strings and DNA Sequences

A string is text. DNA sequences can be represented as strings in Python.

This notebook gives practice with string methods using simple DNA examples.

## 1. Create a DNA sequence

```python
dna = "ATGCGTACCGTA"
dna
```

## 2. Find the sequence length

```python
len(dna)
```

## 3. Count bases

```python
a_count = dna.count("A")
t_count = dna.count("T")
g_count = dna.count("G")
c_count = dna.count("C")

a_count, t_count, g_count, c_count
```

## 4. Calculate GC content

```python
gc_count = dna.count("G") + dna.count("C")
gc_percent = gc_count / len(dna) * 100
gc_percent
```

## 5. Replace letters to make a simple RNA sequence

RNA uses `U` instead of `T`.

```python
rna = dna.replace("T", "U")
rna
```

## 6. Slice part of a sequence

Slicing selects part of a string. The code below selects the first three bases.

```python
dna[0:3]
```
