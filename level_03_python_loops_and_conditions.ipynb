# Level 3: Python Loops and Conditions

Loops repeat an action. Conditions let Python make decisions.

These are useful when you need to process many genes, samples, or measurements.

## 1. A simple for loop

```python
genes = ["TP53", "BRCA1", "EGFR", "BRAF"]

for gene in genes:
    print(gene)
```

## 2. Use a condition

```python
read_depth = 45

if read_depth >= 50:
    print("Pass")
else:
    print("Fail")
```

## 3. Combine loops and conditions

```python
read_depths = [80, 20, 55, 40, 100]

for depth in read_depths:
    if depth >= 50:
        print(depth, "Pass")
    else:
        print(depth, "Fail")
```

## 4. Store results in a new list

```python
passed_depths = []

for depth in read_depths:
    if depth >= 50:
        passed_depths.append(depth)

passed_depths
```

## 5. Count matching items

```python
high_impact_variants = ["HIGH", "LOW", "MODERATE", "HIGH", "HIGH"]
count_high = 0

for impact in high_impact_variants:
    if impact == "HIGH":
        count_high = count_high + 1

count_high
```
