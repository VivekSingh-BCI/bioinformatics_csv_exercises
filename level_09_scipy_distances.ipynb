# Level 9: SciPy Distances

Distances measure how similar or different two numerical profiles are.

In bioinformatics, distances can be used to compare samples, genes, or microbial communities.

```python
import numpy as np
from scipy.spatial import distance
```

## 1. Create two sample profiles

```python
sample_a = np.array([120, 45, 88, 200])
sample_b = np.array([130, 50, 80, 190])

sample_a, sample_b
```

## 2. Calculate Euclidean distance

Euclidean distance is a common straight-line distance between two numerical profiles.

```python
distance.euclidean(sample_a, sample_b)
```

## 3. Calculate cityblock distance

Cityblock distance adds the absolute differences between values.

```python
distance.cityblock(sample_a, sample_b)
```

## 4. Calculate distances between several samples

```python
samples = np.array([
    [120, 45, 88, 200],
    [130, 50, 80, 190],
    [20, 150, 30, 40]
])

distance_matrix = distance.squareform(distance.pdist(samples, metric="euclidean"))
distance_matrix
```

## 5. Add sample names

```python
sample_names = ["S01", "S02", "S03"]

for i, name in enumerate(sample_names):
    print(name, distance_matrix[i])
```
