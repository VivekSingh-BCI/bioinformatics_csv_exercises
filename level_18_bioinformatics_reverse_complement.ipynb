# Level 18: DNA Reverse Complement

DNA has complementary bases:

- A pairs with T
- C pairs with G

The reverse complement is often used when working with DNA sequences.

## 1. Complement one base

```python
base = "A"

if base == "A":
    complement = "T"
elif base == "T":
    complement = "A"
elif base == "C":
    complement = "G"
elif base == "G":
    complement = "C"

complement
```

## 2. Use a dictionary for complements

```python
complement_map = {"A": "T", "T": "A", "C": "G", "G": "C"}
complement_map["A"]
```

## 3. Complement a full sequence

```python
dna = "ATGCCGTA"
complement = ""

for base in dna:
    complement = complement + complement_map[base]

complement
```

## 4. Reverse a sequence

```python
dna[::-1]
```

## 5. Write a reverse complement function

```python
def reverse_complement(dna):
    complement_map = {"A": "T", "T": "A", "C": "G", "G": "C"}
    complement = ""
    for base in dna:
        complement = complement + complement_map[base]
    return complement[::-1]

reverse_complement("ATGCCGTA")
```
