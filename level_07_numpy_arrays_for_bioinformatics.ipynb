# Level 7: NumPy Arrays for Bioinformatics

`NumPy` is a Python library for numerical data.

SciPy uses NumPy arrays, so this notebook gives basic practice before using SciPy.

```python
import numpy as np
```

## 1. Create an array

```python
expression = np.array([120, 135, 98, 150, 170])
expression
```

## 2. Calculate summary statistics

```python
expression.mean()
```

```python
expression.min(), expression.max()
```

## 3. Apply calculations to every value

NumPy can apply one calculation to the whole array.

```python
expression_plus_10 = expression + 10
expression_plus_10
```

## 4. Boolean filtering

```python
high_expression = expression[expression > 130]
high_expression
```

## 5. A two-dimensional array

Rows can represent genes and columns can represent samples.

```python
expression_matrix = np.array([
    [120, 135, 98],
    [45, 50, 42],
    [200, 180, 210]
])

expression_matrix
```

```python
row_means = expression_matrix.mean(axis=1)
row_means
```
